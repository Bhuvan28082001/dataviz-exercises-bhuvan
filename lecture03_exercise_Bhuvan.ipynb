## Task — Multi-Series Line Chart with Highlight

**What to build:** A line chart showing CO2 emissions over time for **all Asian countries** in the dataset, with one country highlighted.

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python

# Task — Multi-series line with highlight

import pandas as pd
import plotly.graph_objects as go

# Load the dataset
df = pd.read_csv('C:\\Users\\ACER\\OneDrive\\Documents\\GitHub\\data-viz-class-material-w2026\\data\\co2_emissions.csv')

# Select all Asian countries
asia_df = df[df['Region'] == 'Asia'].copy()

# Country to highlight
highlight_country = 'China'
highlight_color = '#E76F51'

# Create the figure
fig = go.Figure()

# Add all countries except China in light grey
for country in asia_df['Country'].unique():
    if country != highlight_country:
        country_df = asia_df[asia_df['Country'] == country].sort_values('Year')

        fig.add_trace(go.Scatter(
            x=country_df['Year'],
            y=country_df['CO2_Mt'],
            mode='lines',
            name=country,
            line=dict(
                color='#EEEEEE',
                width=1.5
            ),
            hovertemplate=(
                '<b>' + country + '</b><br>'
                'Year: %{x}<br>'
                'CO₂ emissions: %{y:,.1f} Mt'
                '<extra></extra>'
            ),
            showlegend=False
        ))

# Add China as the highlighted line
china_df = asia_df[
    asia_df['Country'] == highlight_country
].sort_values('Year')

fig.add_trace(go.Scatter(
    x=china_df['Year'],
    y=china_df['CO2_Mt'],
    mode='lines',
    name=highlight_country,
    line=dict(
        color=highlight_color,
        width=3.5
    ),
    hovertemplate=(
        '<b>China</b><br>'
        'Year: %{x}<br>'
        'CO₂ emissions: %{y:,.1f} Mt'
        '<extra></extra>'
    ),
    showlegend=False
))

# Label China directly at the end of its line
last_year = china_df['Year'].iloc[-1]
last_emissions = china_df['CO2_Mt'].iloc[-1]

fig.add_annotation(
    x=last_year,
    y=last_emissions,
    text='China',
    showarrow=False,
    xshift=28,
    font=dict(
        size=13,
        color=highlight_color
    ),
    xanchor='left'
)

# Clean chart layout
fig.update_layout(
    title=dict(
        text="China's CO₂ emissions rose sharply from 2000 to 2022",
        x=0.02,
        xanchor='left',
        font=dict(size=20)
    ),
    xaxis_title='Year',
    yaxis_title='CO₂ emissions (million tonnes)',
    template='simple_white',
    showlegend=False,
    plot_bgcolor='white',
    paper_bgcolor='white',
    margin=dict(l=70, r=100, t=90, b=60),
    hovermode='x unified'
)

# Remove chart borders and heavy gridlines
fig.update_xaxes(
    showgrid=False,
    showline=False,
    zeroline=False,
    dtick=5
)

fig.update_yaxes(
    showgrid=True,
    gridcolor='#F2F2F2',
    gridwidth=1,
    showline=False,
    zeroline=False
)

# Display the chart
fig.show()
```
